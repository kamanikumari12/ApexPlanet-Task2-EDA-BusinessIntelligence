```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_excel(
    "ApexPlanet_DataAnalytics_Dataset(1).xlsx"
 )

# Data cleaning
df['Age'] = df['Age'].fillna(df['Age'].median())
df['City'] = df['City'].fillna('Unknown')
df['Order_Date'] = pd.to_datetime(df['Order_Date'])


# =====================================================
# Q1: Top 5 Products
# =====================================================

q1 = (
    df.groupby('Product')['Total_Sales']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print("\nQ1 - Top 5 Products")
print(q1)

plt.figure(figsize=(8,5))
q1.plot(kind='bar')
plt.title('Top 5 Products by Total Sales')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# =====================================================
# Q2: Sales by Category
# =====================================================

q2 = (
    df.groupby('Category')['Total_Sales']
    .sum()
    .sort_values(ascending=False)
)

print("\nQ2 - Sales by Category")
print(q2)

plt.figure(figsize=(8,5))
q2.plot(kind='bar')
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# =====================================================
# Q3: Monthly Sales Trend
# =====================================================

q3 = (
    df.groupby(
        df['Order_Date'].dt.to_period('M')
    )['Total_Sales']
    .sum()
)

print("\nQ3 - Monthly Sales")
print(q3)

plt.figure(figsize=(11,5))
q3.plot(kind='line', marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Monthly Sales')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# =====================================================
# Q4: Sales by City
# =====================================================

q4 = (
    df.groupby('City')['Total_Sales']
    .sum()
    .sort_values(ascending=False)
)

print("\nQ4 - Sales by City")
print(q4)

plt.figure(figsize=(9,5))
q4.plot(kind='bar')
plt.title('Total Sales by City')
plt.xlabel('City')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# =====================================================
# Q5: Sales by Gender
# =====================================================

q5 = (
    df.groupby('Gender')['Total_Sales']
    .sum()
    .sort_values(ascending=False)
)

print("\nQ5 - Sales by Gender")
print(q5)

plt.figure(figsize=(7,5))
q5.plot(kind='bar')
plt.title('Total Sales by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# =====================================================
# Q6: Top 5 Customers
# =====================================================

q6 = (
    df.groupby(
        ['Customer_ID', 'Customer_Name']
    )['Total_Sales']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print("\nQ6 - Top 5 Customers")
print(q6)

top_customers = (
    df.groupby('Customer_Name')['Total_Sales']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

plt.figure(figsize=(9,5))
top_customers.plot(kind='bar')
plt.title('Top 5 Customers by Total Sales')
plt.xlabel('Customer')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# =====================================================
# Q7: Quantity vs Total Sales
# =====================================================

q7 = (
    df.groupby('Quantity')['Total_Sales']
    .sum()
    .sort_index()
)

print("\nQ7 - Quantity vs Total Sales")
print(q7)

plt.figure(figsize=(8,5))
plt.scatter(
    df['Quantity'],
    df['Total_Sales'],
    alpha=0.5
)

plt.title('Quantity vs Total Sales')
plt.xlabel('Quantity')
plt.ylabel('Total Sales')
plt.tight_layout()
plt.show()

correlation = df['Quantity'].corr(
    df['Total_Sales']
)

print(
    "\nCorrelation:",
    round(correlation, 3)
)
```

```python
import pandas as pd
# Replace with your actual file path on your PC
excel_path = r'C:\Users\Kamani Kumari\Downloads\ApexPlanet_DataAnalytics_Dataset (1).xlsx'

df = pd.read_excel(excel_path, sheet_name='Sales_Dataset')
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
print("Successfully loaded dataset! Shape:", df.shape)

# 3. Query 1: Top categories by revenue
q1 = df.groupby('Category')['Total_Sales'].sum().reset_index().sort_values(by='Total_Sales', ascending=False)
print(q1)
df['YearMonth'] = df['Order_Date'].dt.to_period('M')
q2 = df.groupby('YearMonth').agg(
    Total_Orders=('Order_ID', 'count'),
    Total_Revenue=('Total_Sales', 'sum')
).reset_index()

# Query 3: Top 5 Cities by Total Sales (Defines q3)
q3 = df.groupby('City')['Total_Sales'].sum().reset_index().sort_values(by='Total_Sales', ascending=False).head(5)
# Query 4: Average Order Value by Gender
q4 = df.groupby('Gender')['Total_Sales'].mean().reset_index()

# Query 5: Top 5 Products by Quantity Sold
q5 = df.groupby('Product')['Quantity'].sum().reset_index().sort_values(by='Quantity', ascending=False)

print("Q1 Top Categories:\n", q1)
print("\nQ2 Monthly Trend Head:\n", q2.head())
print("\nQ3 Top Cities:\n", q3)
print("\nQ4 AOV by Gender:\n", q4)
print("\nQ5 Top Products:\n", q5)
```
